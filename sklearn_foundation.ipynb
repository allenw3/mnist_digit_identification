```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
```

```python
mnist = fetch_openml("mnist_784", version=1) #mnist is handwritten digits dataset!
x , y = mnist['data'], mnist['target'].astype('int')
x = x.to_numpy()
y = y.to_numpy() #make sure these are numpy arrays, not pandas dataframes!
```

```python
plt.close()
plt.imshow(x[1].reshape(28, 28), cmap = 'grey') #reshape to grid
plt.show()
plt.close()
print(len(mnist))
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 40) 
# x is digit images, y is digit labels
#y_train is  20% of the data used to test the model (test size 0.2) (rest is for training)
#random_state = 40 - shuffling for reproducibility
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter = 1000) 
#training a Logistic Regression model on the training data - 1000 iterations max to converge on a solution
model.fit(x_train, y_train)
print("Finished!")
```

```python
from sklearn.metrics import accuracy_score
y_pred = model.predict(x_test) #uses .predict() to guess labels for the test images
print("Accuracy:", accuracy_score(y_test, y_pred), "Yippee!") #how accurate is y_pred to y_test?
```
